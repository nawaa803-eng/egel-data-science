```python
# ada data yg kosong
# ada data yang duplikat
# jika kosong pada feature (x) === jika numerik === hitung skew()
# tentukan fillna pake mean atau mediam

# jika absolute skew < 0.5 === mean
# jika absolute skew = 0.5 === median

#jika kosong pada feature (x) ==== jika kategorikal
# fillna dengan mode

# jika kosong pada target (y) === hapus baris data
# hapus data yg duplikat

print(abs(df["berat"].skew()))
print(abs(df["diameter"].skew()))
print(abs(df["kadar_air"].skew()))

# jumlah duplikat
df = df.dropna(subset=["grade"])
```

```python
df.isna().sum()
```

```python
# isi kolom numerik dengan mean
df["berat"] = df["berat"].fillna(df["berat"].mean())
df["diameter"] = df["diameter"].fillna(df["diameter"].mean())
df["kadar_air"] = df["kadar_air"].fillna(df["kadar_air"].mean())

# isi kolom kategori dengan mode
df["jenis_kentang"] = df["jenis_kentang"].fillna()
```
